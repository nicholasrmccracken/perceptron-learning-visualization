```python
# PART A

import matplotlib.pyplot as plt
import numpy as np

def plot_points(x, d, title):
    plt.figure()
    for i in range(len(d)):
        if d[i] == 0:
            plt.scatter(x[i, 0], x[i, 1], marker='x', color='red')
    for i in range(len(d)):
        if d[i] == 1:
            plt.scatter(x[i, 0], x[i, 1], marker='o', color='blue')

    # Empty scatters for legend
    plt.scatter([], [], marker='x', color='red', label='C1')
    plt.scatter([], [], marker='o', color='blue', label='C2')

    plt.title(title)
    plt.xlabel('x1')
    plt.ylabel('x2')
    plt.legend()
    plt.grid()
    plt.show()

x = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])
d = np.array([0, 0, 1, 1])  # C1 = 0, C2 = 1

plot_points(x, d, "(PART A) Training Samples")
```

```python
# PART B
def plr(x, w, d, eta, L):
    w = w.copy()
    x = np.hstack((np.ones((x.shape[0], 1)), x))  # Add a column of 1s for bias
    n_samples = x.shape[0]

    for iteration in range(L):
        print(f"\nIteration {iteration + 1}")
        for i in range(n_samples):
            y = 1 if np.dot(w, x[i]) >= 0 else 0

            if y != d[i]:
                w = w + eta * (d[i] - y) * x[i]
                print(f"Updated weights: {w}")

    return w
```

```python
# PART C
w = np.array([0.0, 0.0, 0.0])  # 3 weights: bias, x1, x2
eta = 0.5
L = 10

final_weights = plr(x, w, d, eta, L)
print("\n(PART C) Final weights:", final_weights)
```

```python
# PART E

d = np.array([1, 0, 0, 1])

plot_points(x, d, "(PART E) Training Samples")

final_weights = plr(x, w, d, eta, L)
print("\n(PART E) Final weights:", final_weights)
```
